## Conceptos clave (antes de los GRANT)

En Databricks, la jerarquía de objetos es:

```
CATALOG
 └── SCHEMA (database)
      └── TABLE / VIEW
```

Regla importante:
👉 **Si un usuario no tiene permiso sobre el catálogo o el esquema, no podrá acceder a la tabla**, aunque tenga `SELECT`.

# Permiso para usar el catálogo

🔹 Permite al usuario **ver y usar** el catálogo `maintenance_iot`.

Sin este permiso:
❌ No puede acceder a ningún esquema ni tabla dentro del catálogo.

````sql

GRANT USE CATALOG ON CATALOG maintenance_iot
TO `dev.cristian.bohorquez@gmail.com`;
````

# Permiso para usar un esquema (database)

🔹 Permite **acceder al esquema `bronze`**.

📌 Es obligatorio para:
- Consultar tablas
- Crear objetos (si tiene permisos adicionales)

````sql
GRANT USE SCHEMA ON SCHEMA maintenance_iot.bronze
TO `dev.cristian.bohorquez@gmail.com`;
````

# Permiso de lectura sobre una tabla

🔹 Concede permiso **solo de lectura** sobre la tabla `sensor_events`.

📌 Importante:
- Este permiso **no funciona solo**.
- El usuario también necesita `USE CATALOG` y `USE SCHEMA`.

````sql
GRANT SELECT ON TABLE maintenance_iot.bronze.sensor_events
TO `dev.cristian.bohorquez@gmail.com`;
````

# Permisos para el grupo Desarrolladores_Silver

````sql
GRANT USE CATALOG ON CATALOG maintenance_iot
TO `Desarrolladores_Silver`;

GRANT USE SCHEMA ON SCHEMA maintenance_iot.silver
TO `Desarrolladores_Silver`;

GRANT CREATE ON SCHEMA maintenance_iot.silver
TO `Desarrolladores_Silver`;

GRANT USE SCHEMA ON SCHEMA maintenance_iot.bronze
TO `Desarrolladores_Silver`;

GRANT SELECT ON ALL TABLES IN SCHEMA maintenance_iot.bronze
TO `Desarrolladores_Silver`;
````
